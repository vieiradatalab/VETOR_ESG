# O que significa “parecido” em dados ESG?
Quando alguém pergunta: **“Quais países são mais parecidos com o Brasil em termos de indicadores ESG?”**  
A resposta mais comum costuma vir em forma de **médias**, **rankings** ou **índices agregados**.  
Mas existe um problema metodológico aqui.
**ESG não é uma variável**. É um **conjunto de variáveis ao mesmo tempo**.

Indicadores <u>ambientais</u>, <u>sociais</u> e de <u>governança</u> formam um **sistema multidimensional**.  
Quando tentamos comparar países apenas olhando <u>médias</u> ou <u>rankings</u>, estamos simplificando demais um problema que é naturalmente mais complexo.  
Medidas como **média**, **mediana**, **variância** e **desvio padrão** são extremamente úteis mas, elas respondem a perguntas específicas:
 - <u><span style="color:blue">Qual é o valor típico de um indicador?</span></u>
 - <u><span style="color:blue">Quanto os valores variam?</span></u>
 - <u><span style="color:blue">Um país está acima ou abaixo da média?</span></u>

Essas métricas ajudam a entender **cada indicador isoladamente**.
Mas elas não respondem diretamente à pergunta: **Quais países são mais parecidos considerando todos os indicadores ao mesmo tempo?** Aqui surge uma mudança de perspectiva interessante. Em vez de pensar em cada indicador separadamente, podemos representar cada país como um **vetor de indicadores ESG**.

Por exemplo, o Brasil:
- CO<sup>2</sup> per capita
- Energia renovável
- Expectativa de vida
- Mortalidade infantil
- Controle da corrupção
- Efetividade do governo

> **Nota:** este experimento tem finalidade didática; os indicadores foram selecionados intencionalmente para ilustrar a abordagem.

Outro país terá o mesmo conjunto de dimensões. Isso nos leva a uma abordagem diferente: **medir a distância entre esses vetores**. Se dois países têm vetores muito próximos, eles apresentam **perfis ESG semelhantes**. Se a distância é grande, os perfis são bastante diferentes.

Uma das formas mais simples de fazer isso é usar **distância euclidiana**, uma medida clássica em análise multivariada. Nesse caso, comparar países vira literalmente um problema de **geometria de dados**. Mas existe um detalhe importante: antes de calcular distâncias, os indicadores precisam ser **padronizados**, porque cada variável possui escala diferente (anos, toneladas, índices etc.).
Sem essa etapa, a comparação pode ficar completamente distorcida.

Essa abordagem levanta uma pergunta interessante: Talvez a discussão sobre ESG não devesse se limitar apenas a **quem está melhor no ranking**. Talvez também devêssemos perguntar: **quais países realmente se parecem entre si em termos de estrutura ESG?**

Para explorar essa ideia na prática, preparei um pequeno experimento com dados do World Development Indicators do Banco Mundial.  
O código completo e o notebook estão disponíveis neste repositório no GitHub.

```python
# Cria session com retry/backoff
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

def build_session() -> requests.Session:
    """ Session com retry/backoff para tornar o ETL robusto a:
    - timeouts intermitentes
    - 429/5xx do servidor
    """
    session = requests.Session()

    retry = Retry(
        total=8,
        connect=8,
        read=8,
        backoff_factor=1.2,  # 1.2s, 2.4s, 4.8s...
        status_forcelist=[429, 500, 502, 503, 504],
        allowed_methods=["GET"],
        raise_on_status=False,
        respect_retry_after_header=True,
    )

    adapter = HTTPAdapter(max_retries=retry, pool_connections=10, pool_maxsize=10)
    session.mount("https://", adapter)
    session.mount("http://", adapter)
    return session

SESSION = build_session()
```

```python
INDICATORS = {
  "E": {
    "EN.ATM.CO2E.PC": "CO2 per capita",
    "EG.FEC.RNEW.ZS": "Renewable energy consumption (%)"
  },
  "S": {
    "SP.DYN.LE00.IN": "Life expectancy at birth (years)",
    "SP.DYN.IMRT.IN": "Infant mortality rate (per 1,000 live births)"
  },
  "G": {
    "CC.EST": "Control of Corruption",
    "GE.EST": "Government Effectiveness"
  }
}

YEAR = 2019
TARGET_COUNTRY = "Brazil"
TOP_N = 10
```

```python
def download_data(indicators, year):
    available_indicators = []
    skipped_indicators = []
    data_frames = []

    for category, ind_dict in indicators.items():
        for code, desc in ind_dict.items():
            try:
                df = wb.download(indicator=code, country="all", start=year, end=year)
                df = df.reset_index().pivot(index="country", columns="year", values=code)
                df.columns = [f"{code}_{year}"]
                data_frames.append(df)
                available_indicators.append(code)
                logging.info(f"OK: {code} ({desc})")
            except Exception as e:
                logging.warning(f"SKIP: {code} ({desc}) -> {e}")
                skipped_indicators.append(code)

    combined_df = pd.concat(data_frames, axis=1, join="outer") if data_frames else pd.DataFrame()
    return combined_df, available_indicators, skipped_indicators
```

```python
raw_df, available, skipped = download_data(INDICATORS, YEAR)

print("Shape (raw):", raw_df.shape)
raw_df.head()
```

```python
def handle_missing(data, threshold=0.4):
    # Remove países com mais de threshold de valores faltantes
    missing_pct = data.isnull().mean(axis=1)
    data = data[missing_pct <= threshold].copy()

    # Imputa o restante com mediana por coluna
    data = data.fillna(data.median(numeric_only=True))
    return data

clean_df = handle_missing(raw_df, threshold=0.4)

print("Shape (clean):", clean_df.shape)
clean_df.head()
```

```python
def calculate_similarities(X, target_country):
    if target_country not in X.index:
        raise ValueError(
            f"País alvo '{target_country}' não encontrado. "
            f"Exemplos disponíveis: {list(X.index[:15])}"
        )

    target_vec = X.loc[target_country].values.reshape(1, -1)

    euclidean_dist = np.linalg.norm(X.values - target_vec, axis=1)
    cosine_sim = cosine_similarity(X.values, target_vec).flatten()

    results = pd.DataFrame({
        "country": X.index,
        "euclidean_distance": euclidean_dist,
        "cosine_similarity": cosine_sim
    }).sort_values("cosine_similarity")

    return results

sim = calculate_similarities(X, TARGET_COUNTRY)
sim.head(15)
```

```python
top_euclidean = sim.nsmallest(TOP_N, "euclidean_distance")
top_cosine = sim.nlargest(TOP_N, "cosine_similarity")

print(f"Top {TOP_N} por Distância Euclidiana (menor = mais parecido):")
display(top_euclidean)

print(f"Top {TOP_N} por Similaridade Cosseno (maior = mais parecido):")
display(top_cosine)
```

```python
def pca_svd(X_values, n_components=2):
    X_centered = X_values - X_values.mean(axis=0, keepdims=True)
    U, s, Vt = np.linalg.svd(X_centered, full_matrices=False)
    Z = U[:, :n_components] * s[:n_components]
    return Z

Z = pca_svd(X.values, n_components=2)
Z.shape
```

```python
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

def plot_pca(Z, countries, target_country, highlight_countries, zoom_xlim=(-1, 1), zoom_ylim=(-1, 1)):
    plt.figure(figsize=(10, 8))
    ax = plt.gca()

    highlight_set = set([target_country] + list(highlight_countries))
    is_highlight = [c in highlight_set for c in countries]

    sns.scatterplot(x=Z[:, 0], y=Z[:, 1], hue=is_highlight, palette="viridis", ax=ax)

    for i, c in enumerate(countries):
        if c in highlight_set:
            ax.annotate(c, (Z[i, 0], Z[i, 1]), fontsize=9)

    ax.set_title("Projeção PCA (SVD): Brasil e países mais similares (ESG)")
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    ax.legend(title="Destaque")

    # Inset (zoom)
    axins = inset_axes(ax, width="40%", height="40%", loc="upper right")
    sns.scatterplot(x=Z[:, 0], y=Z[:, 1], hue=is_highlight, palette="viridis", ax=axins, legend=False)

    axins.set_xlim(*zoom_xlim)
    axins.set_ylim(*zoom_ylim)
    axins.set_title("Zoom", fontsize=9)
    axins.set_xlabel("")
    axins.set_ylabel("")

    plt.show()

highlight = top_euclidean["country"].tolist()
plot_pca(Z, list(X.index), TARGET_COUNTRY, highlight, zoom_xlim=(-1, 1), zoom_ylim=(-1, 1))
```
